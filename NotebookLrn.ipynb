```python
#execrise
limit= float(input("Enter total monthly data limit in (GB): "))
Data_used = float(input("Data used so far (GB): "))

remaining_data= limit-Data_used

usage_percentage= round((Data_used/limit)*100,2)

print("Remaining data is", remaining_data ,"GB")
print(f"You have used up to {usage_percentage}GB of your monthly limit")

if usage_percentage>=80:
    print('Warning: High usage, consider upgrading your plan.')
else:
    print('Enjoy Browsing')
```

```python
original_price= float(input("Enter the original price (rs): "))
discount_percent = int(input("Enter the discount percentage (%): "))

discount_amount = original_price*(discount_percent/100)
print(f"Your discount applied is {discount_amount} %")
final_price= original_price-discount_amount
print(f"Final price after discount {final_price}rs/-")
```
